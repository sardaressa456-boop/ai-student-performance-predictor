# AI Student Performance Predictor — Exploratory Data Analysis
This notebook explores the UCI Student Performance dataset (Mathematics) to inspect distributions, evaluate correlations, and prepare features for machine learning modeling.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load dataset
data_path = Path("../data/student-mat.csv")
df = pd.read_csv(data_path, sep=";")
print(f"Shape: {df.shape}")
df.head()
```

```python
# Data statistics and missing values
print("Missing values:", df.isnull().sum().sum())
df.describe()
```

```python
# Grade correlation
numeric_df = df.select_dtypes(include="number")
numeric_df.corr()["G3"].sort_values(ascending=False)
```

```python
# Visualization: Grade Distribution
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 4))
sns.histplot(df["G3"], bins=21, kde=True, color="royalblue")
plt.title("Final Grade Distribution (G3)")
plt.xlabel("Grade (0-20)")
plt.ylabel("Count")
plt.show()
```
